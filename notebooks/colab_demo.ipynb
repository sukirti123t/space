## Alternative: quick headless test of the core modules (no UI)
Use this if you just need to verify `detect.py` and `orbit.py` work in the
Colab runtime without launching the full dashboard.

```python
from detect import DebrisDetector, simulate_synthetic_frame

img = simulate_synthetic_frame(seed=42)
detector = DebrisDetector()
annotated, detections = detector.detect(img, conf_threshold=0.2)
print(f"Found {len(detections)} candidate objects")

import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
plt.imshow(annotated)
plt.axis('off')
plt.show()
```

```python
import datetime as dt
from orbit import fetch_tle, analyze_conjunction

name_a, l1_a, l2_a = fetch_tle(44804)  # Cartosat-3
name_b, l1_b, l2_b = fetch_tle(40269)  # PSLV debris

start = dt.datetime.now(dt.timezone.utc)
end = start + dt.timedelta(hours=6)

result = analyze_conjunction(name_a, l1_a, l2_a, name_b, l1_b, l2_b, start, end)
print(result)
```
